```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

Dataset Link: https://www.kaggle.com/datasets/blastchar/telco-customer-churn

```python
df = pd.read_csv('data/WA_Fn-UseC_-Telco-Customer-Churn.csv',low_memory=False)
```

```python
df.head()
```

```python
df.info()
```

```python
 df[df['TotalCharges'] == ' ']
```

TotalCharges sütununun object tipinde olmasından şüphelenilerek veri incelendiğinde, tenure (müşteri olma süresi) değeri 0 olan 11 yeni müşterinin toplam fatura değerlerinin sisteme 0 yerine 'boşluk' (space) olarak kaydedildiği tespit edilmiştir.

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'],errors='coerce')
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
for col in df.select_dtypes(include=[object]).columns:
    print(f'{col}')
    print(df[col].unique())
    print('\n')
```

```python
df.drop('customerID',axis=1,inplace=True)
```

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df,x='Churn',palette='viridis')
plt.title('Müşteri Kaybı Dağılımı')
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.countplot(data=df,x='gender',hue='Churn',palette='Set2')
plt.title('Cinsiyete Göre Müşteri Kaybı Grafiği')
plt.xlabel('Cinsiyet')
plt.ylabel('Müşteri Sayısı')
plt.show()
```

```python
df.drop('gender',axis=1,inplace=True)
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df,x='Contract',hue='Churn',palette='magma')
plt.title('Kontrat Süresine Göre Müşteri Kaybı Grafiği')
plt.xlabel('Kontrat')
plt.ylabel('Müşteri Sayısı')
plt.show()
```
